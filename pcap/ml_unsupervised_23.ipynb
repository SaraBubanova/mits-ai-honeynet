```python
columns_to_drop = ["id", "src_ip", "src_port", "dst_port", "src_mac", "src_oui", "dst_ip", "dst_mac", "dst_oui", 
                   "application_name", "application_category_name", "client_fingerprint", "server_fingerprint", 
                   "requested_server_name", "user_agent", "content_type", "bidirectional_first_seen_ms", "bidirectional_last_seen_ms", 
                   "src2dst_first_seen_ms", "src2dst_last_seen_ms", "dst2src_first_seen_ms", "dst2src_last_seen_ms", "tunnel_id", 
                   "application_confidence", "splt_direction", "splt_ps", "splt_piat_ms"]

# Drop STRING columns 
df = df.drop([col for col in columns_to_drop if col in df.columns])

# Find the variance for each feature
nunique_df = df.select(pl.all().n_unique())

constant_cols = [
  col for col, val in zip(nunique_df.columns, nunique_df.row(0))
  if val <= 1
]

df = df.drop(constant_cols)

print(f"Dropped {len(columns_to_drop)} columns composed by STRINGS and {len(constant_cols)} columns with close to zero variance")

df.head()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import numpy as np

# Numpy for df
X = df.to_numpy()

# # Scaler
# scaler = StandardScaler()
# X_scaled = scaler.fit_transform(X)

from sklearn.preprocessing import RobustScaler
scaler = RobustScaler()
X_scaled = scaler.fit_transform(X)

X_scaled = np.clip(X_scaled, -5, 5)

# PCA
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

print(f"PCA Dimensions: {X_pca.shape[1]}")
```

```python
import matplotlib.pyplot as plt
plt.style.use("ggplot") 
plt.plot(pca.explained_variance_, marker='o')
plt.xlabel("Eigenvalue number")
plt.ylabel("Eigenvalue size")
plt.title("Scree Plot")
```

```python
cumvar = np.cumsum(pca.explained_variance_ratio_)
n_components = np.argmax(cumvar >= 0.95) + 1
print(f"Components for 95% variance: {n_components}")
print(f"Components for 90% variance: {np.argmax(cumvar >= 0.90) + 1}")
```

```python
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler
import polars as pl
import numpy as np

pca = PCA(n_components=n_components)
X_pca = pca.fit_transform(X_scaled)

projection = TSNE(
  n_components=2,
  perplexity=40,
  init="pca",
  learning_rate="auto",
  random_state=42,
).fit_transform(X_pca)

# Plot
plt.figure(figsize=(8, 6))
plt.scatter(
  projection[:, 0],
  projection[:, 1],
  s=15,
  alpha=0.7,
)
plt.title("t-SNE Projection")
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")
plt.show()
```

```python
from sklearn.neighbors import NearestNeighbors
from kneed import KneeLocator
import numpy as np

# Double Dimension refer to link
min_samples = X_pca.shape[1] * 2 

nbrs = NearestNeighbors(n_neighbors=min_samples).fit(X_pca)
distances, _ = nbrs.kneighbors(X_pca)

k_distances = np.sort(distances[:, -1])

# Detect knee
kl = KneeLocator(range(len(k_distances)), k_distances, curve="concave", direction="increasing")
optimal_eps = k_distances[kl.knee]

print(f"Optimal Epsilon: {optimal_eps:.4f}")
print(f"Min Dist: {k_distances.min()}")
print(f"50 percentile Dist: {np.percentile(k_distances, 50)}")
print(f"95 percentile Dis: {np.percentile(k_distances, 95)}")
print(f"Max Dist: {k_distances.max()}")
```

```python
from sklearn.cluster import DBSCAN

for pct in [50, 60, 70, 75, 80, 85, 90, 95]:
  eps_test = np.percentile(k_distances, pct)
  test_clusters = DBSCAN(eps=eps_test, min_samples=min_samples).fit_predict(X_pca)
  n = len(set(test_clusters)) - (1 if -1 in test_clusters else 0)
  noise = (test_clusters == -1).sum()
  print(f"pct={pct}  eps={eps_test:.4f}  clusters={n}  noise={noise/len(test_clusters)*100:.1f}%")
```
